# Train Teacher

```python
import WiDepthLite as WD
from DatasetLite import MyDataLoaderLite as DL
from TrainerLite import ModelTrainer as MT
```

```python
data_loader = DL()
```

```python
data_loader.load('../../dataset/Tanaka_Toy/')
```

```python
train_loader, valid_loader, test_loader = data_loader.gen_loaders()
```

```python
teacher = WD.Teacher(device='cuda')
```

```python
import torch
trainer = MT(model=teacher, dataloader={'train':train_loader, 'valid':valid_loader, 'test':test_loader}, device=torch.cuda.device)
```

```python
trainer.train()
```
