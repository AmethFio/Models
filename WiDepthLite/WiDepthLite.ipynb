```python
import WiDepthLite as WD
from DatasetLite import DataLoaderLite as DL
from TrainerLite import ModelTrainer as MT
import torch
```

```python
data_loader = DL()
```

```python
data_loader.load('../../dataset/Tanaka_Toy/')
```

```python
train_loader, valid_loader, test_loader = data_loader.gen_loaders()
```

```python
teacher = WD.Teacher(device=torch.device('cuda:2'))
```

```python
import torch
trainer = MT(name='Wmodel=teacher, dataloader={'train':train_loader, 'valid':valid_loader, 'test':test_loader}, device=2)
```

```python
trainer.train()
```
