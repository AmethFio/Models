# Imports and load data

```python
import WiDepthLite as WD
from DatasetLite import DataLoaderLite as DL
from WiDepthLite import StudentTrainer as ST
import torch
```

```python
data_loader = DL()
```

```python
data_loader.load('../../dataset/Tanaka_Test/')
```

```python
train_loader, valid_loader, test_loader = data_loader.gen_loaders()
```

# Train Student

```python
trainer = ST(name='Student', notion='shape_test',
             dataloader={'train':train_loader, 'valid':valid_loader, 'test':test_loader}, 
             device=2)
```

```python
trainer.load('../saved/shape_test/', name='Teacher', mode='best')
```

```python
trainer.train()
```

```python
print(getattr(trainer.model, 'imgde').teacher_training)
```

# Test Student

```python
train_loader, valid_loader, test_loader = data_loader.gen_loaders(split_ratio=1)
```

```python
trainer = ST(name='Student', notion='shapedark',
             dataloader={'train':train_loader, 'valid':valid_loader, 'test':test_loader}, 
             device=2)
```

```python
trainer.load('../saved/shape_test/', name='Student', mode='best')
```

```python
trainer.test()
```

```python
trainer.tester.loss_tracker.save()
```
